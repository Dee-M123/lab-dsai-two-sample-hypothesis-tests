In this challenge, we will be using the Pokemon dataset. Before applying statistical methods to this data, let's first examine the data.

Let's start off by looking at the `head` function in the cell below.

The first thing we would like to do is compare the legendary Pokemon to the regular Pokemon. To do this, we should examine the data further. What is the count of legendary vs. non legendary Pokemons?

```python
pokemon['Legendary'].value_counts()
```

Compute the mean and standard deviation of the total points for both legendary and non-legendary Pokemon.

```python
pokemon.groupby('Legendary')['Total'].agg(['mean', 'std'])
```

The computation of the mean might give us a clue regarding how the statistical test may turn out; However, it certainly does not prove whether there is a significant difference between the two groups.

In the cell below, use the `ttest_ind` function in `scipy.stats` to compare the the total points for legendary and non-legendary Pokemon. Since we do not have any information about the population, assume the variances are not equal.

```python
from scipy.stats import ttest_ind

# Two groups: legendary and non legendary

legendary = pokemon[pokemon['Legendary'] == True]['Total']
non_legendary = pokemon[pokemon['Legendary'] == False]['Total']

# H_0: legendary = on legendary
# H_0: legendary != on legendary

# independent test
t_stat, p_value = ttest_ind(legendary, non_legendary, alternative="two-sided", equal_var=False)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

What do you conclude from this test? Write your conclusions below.

```python
# The extremely small p-value also indicates very strong evidence that the difference between the two groups is not due to random chance
```

How about we try to compare the different types of pokemon? In the cell below, list the types of Pokemon from column `Type 1` and the count of each type.

Since water is the largest group of Pokemon, compare the mean and standard deviation of water Pokemon to all other Pokemon.

```python
# Create two groups
water = pokemon[pokemon['Type 1'] == 'Water']['Total']
non_water = pokemon[pokemon['Type 1'] != 'Water']['Total']

# mean and standard deviation
print("Water Pokémon")
print("Mean:", water.mean())
print("Standard Deviation:", water.std())

print("\nNon_Water Pokémon")
print("Mean:", non_water.mean())
print("Standard Deviation:", non_water.std())
```

Perform a hypothesis test comparing the mean of total points for water Pokemon to all non-water Pokemon. Assume the variances are equal.

```python
# H_0: water = others
# H_1: water != others

t_stat, p_value = ttest_ind(water, non_water,alternative="two-sided", equal_var=True)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

Write your conclusion below.

```python
# failed to reject H_0; no statistically significant difference in the mean total points between Water Pokémon and non-Water Pokémon
```

# Challenge 2 - Matched Pairs Test

In this challenge we will compare dependent samples of data describing our Pokemon. Our goal is to see whether there is a significant difference between each Pokemon's defense and attack scores. Our hypothesis is that the defense and attack scores are equal. In the cell below, import the `ttest_rel` function from `scipy.stats` and compare the two columns to see if there is a statistically significant difference between them.

```python
from scipy.stats import ttest_rel

attack = pokemon['Attack']
defense = pokemon['Defense']
# H_0: attack = defense
# H_1: attack != defense

# Perform paired t-test
t_stat, p_value = ttest_rel(attack, defense, alternative="two-sided")

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

Describe the results of the test in the cell below.

```python
# we reject the null hypothesis that the mean Attack and Defense scores are equal.
```

We are also curious about whether therer is a significant difference between the mean of special defense and the mean of special attack. Perform the hypothesis test in the cell below.

```python
s_attack = pokemon['Sp. Atk']
s_defense = pokemon['Sp. Def']
# H_0: s_attack = s_defense
# H_1: s_attack != s_defense

# paired t-test
t_stat, p_value = ttest_rel(s_attack, s_defense, alternative="two-sided")

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

Describe the results of the test in the cell below.

```python
# fail to reject the null hypothesis
```

As you may recall, a two sample matched pairs test can also be expressed as a one sample test of the difference between the two dependent columns.

Import the `ttest_1samp` function and perform a one sample t-test of the difference between defense and attack. Test the hypothesis that the difference between the means is zero. Confirm that the results of the test are the same.

```python
from scipy.stats import ttest_1samp
# difference between Attack and Defense
diff = attack - defense

#H_0: diff = 0
#H_1: diff != 0

# One-sample t-test (testing if mean difference = 0)
t_stat, p_value = ttest_1samp(diff, 0)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
# we reject the null hypothesis that the mean difference between Attack and Defense is zero
```
